# Insurance Claims Analysis – Coding Challenge (Using Lists)
You are a Data Analyst working for an insurance company. You received raw claim data stored in
lists.

# Dataset (Lists)


policy_ids = [101, 102, 103, 104, 105, 106]


claim_amounts = [5000, 15000, 7000, 20000, 3000, 12000]


claim_status = ["Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved"]


customer_age = [25, 45, 35, 50, 23, 40]

Tasks


Task 1: Total Approved Claim Amount
    
    
Calculate the total claim amount for all Approved claims.

```python
policy_ids = [101, 102, 103, 104, 105, 106]

claim_amounts = [5000, 15000, 7000, 20000, 3000, 12000]

claim_status = ["Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved"]

customer_age = [25, 45, 35, 50, 23, 40]

appor_amt = 0

for i in range(len(claim_status)):
    if claim_status[i] == 'Approved':
        appor_amt = appor_amt + claim_amounts[i]

print("Total claim Amount for all Approved claims:",appor_amt)
```

Task 2: Count Rejected Claims
    
    
Find how many claims are Rejected.

```python
reje_cout = 0

for i in claim_status:
    if i == 'Rejected':
        reje_cout = reje_cout +1

print("Cont of Claims got rejected :",reje_cout)
```

Task 3: High-Value Claims
    
    
Print all policy IDs where claim amount is greater than ₹10,000.

```python
High_order = []

for i in range(len(claim_amounts)):
    if claim_amounts[i] > 10000:
        High_order.append(policy_ids[i])
        
print("All Policy IDs where claim amount is greater than ₹10,000:",High_order)
```

Task 4: Average Age of Approved Customers
    
    
Find the average age of customers whose claims are Approved.

```python
appor_age = []
count = 0

for i in range(len(claim_status)):
    if claim_status[i] == 'Approved':
        count = count+1
        appor_age.append(customer_age[i])

print("The average age of customers whose claims are approved:",sum(appor_age)/count)
```

Task 5: Risk Category Tagging
    
    
Create a new list called risk_category:
    
    

• Age < 30 → "Low Risk"


• Age 30–45 → "Medium Risk"


• Age > 45 → "High Risk"

```python
risk_category = []


for i in customer_age:
    if i < 30:
        risk = "Low Risk"
    elif i >= 30 and i <=45 :
        risk ="Medium Risk"
    else:
        risk = "High Risk"
    risk_category.append(risk)

print("Risk_category:",risk_category)
```
